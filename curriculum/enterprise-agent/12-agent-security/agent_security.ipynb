The incident adviser reads a runbook that attempts indirect prompt injection and data exfiltration. We treat all external content as data and independently constrain context, tools, identity, egress, action, and audit.

**Safety boundary:** detection may fail; policy, scope, sandbox, authorization, budget, approval, and kill controls still contain the run.
