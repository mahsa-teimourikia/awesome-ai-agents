# Agent Orchestration

## Scenario: durable EU incident proposal workflow

An orchestrator routes an incident, parallelizes independent evidence, joins artifacts, invokes a bounded agent synthesis node, checkpoints an approval node, and resumes only on a trusted event. The agent does not own scheduling, authorization, or execution.

![Agent orchestration graph](../../../assets/agent-orchestration.svg)

Graph state, queues, timers, checkpoints, budgets, retries, approval, and recovery are deterministic orchestration concerns. Agents provide bounded reasoning inside selected nodes.

## Practical patterns

Use state machines/graphs for explicit conditional paths; DAGs for dependency-ready parallel work; queues/events/schedules for asynchronous wake-up; checkpoints for crash/approval resume; bounded retries and idempotency for safe recovery; and human approval nodes for exact high-risk proposals.
