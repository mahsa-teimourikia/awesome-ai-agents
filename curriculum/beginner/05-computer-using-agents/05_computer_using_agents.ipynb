# Computer-Using Agents: Web & Multimodal
This notebook demonstrates how agents interact with graphical user interfaces using Accessibility Trees (AXTrees) and Multimodal Vision (OmniParser).

## 1. Accessibility Trees (Playwright)
Sending raw HTML DOMs causes token bloat. SOTA web agents extract semantic AXTrees and assign temporary integer IDs to clickable elements.

```python
# Simulated DOM Extraction
raw_html = """
<div class="header">
  <button id="btn-login" class="bg-blue rounded">Login</button>
  <a href="/about" class="text-gray hidden-mobile">About Us</a>
</div>
"""

def extract_axtree(html: str):
    print("🌳 Parsing HTML to clean Accessibility Tree...")
    # Simulated Playwright AXTree extraction
    return {
        0: {"type": "button", "text": "Login", "dom_id": "btn-login"},
        1: {"type": "link", "text": "About Us", "href": "/about"}
    }

axtree = extract_axtree(raw_html)
print("\nPrompt to LLM:")
print(f"Clickable Elements:\n [0] Button: 'Login'\n [1] Link: 'About Us'")

print("\nAgent Output: click(0)")
print(f"Framework Action: Executing native browser click on DOM element '{axtree[0]['dom_id']}'")
```
