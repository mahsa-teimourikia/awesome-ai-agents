# Microsoft Agent Framework — support escalation workflow

## Why this framework here

Northstar must turn a checkout-support signal into a reviewable Gold-customer escalation. The workflow stages are predictable, while a model can help draft an evidence-backed summary. Microsoft Agent Framework is a reasonable candidate when a team wants agents, function tools, workflow builder/execution, state, and Microsoft-oriented hosting/integration paths.

**Safety boundary:** tools are read-only; the draft is not a customer notification; human review remains application-owned.


## Major features to evaluate

- Agent construction and model/provider configuration
- Function tools and typed tool boundaries
- Workflow builder/execution, routing, and state
- Middleware/observability and hosting integration
- Multi-agent workflow samples when specialization is justified

The framework packages runtime mechanics. Identity, tenant scope, budgets, approval, idempotency, and evaluation still belong in application policy.

## Step-by-step implementation design

1. Validate user/tenant identity and classify the request.
2. Call narrow `get_service_status` and `search_tickets` tools.
3. Store evidence IDs in workflow state, not an unbounded transcript.
4. Let an agent draft a concise escalation with citations.
5. Validate the draft schema and send it to an on-call reviewer; only that reviewer may publish/notify.
6. Trace each workflow stage and evaluate tool path, evidence support, latency, and policy.
