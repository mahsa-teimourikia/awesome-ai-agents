# Framework Landscape: Orchestrators vs. Abstractions
This notebook demonstrates the difference between High-Level Abstractions (CrewAI/AutoGen) and Low-Level Orchestrators (LangGraph/PydanticAI).

## 1. High-Level Abstraction (Pseudo-CrewAI)
Magic frameworks hide the execution loop. You define personas and let them figure it out. Great for demos, terrible for debugging.

```python
class MockCrewAIAgent:
    def __init__(self, role, goal):
        self.role = role
        self.goal = goal

researcher = MockCrewAIAgent(role="Senior Researcher", goal="Find facts about Mars.")
writer = MockCrewAIAgent(role="Lead Writer", goal="Turn facts into a blog post.")

print(f"🧙‍♂️ Initiating Magic Crew Framework...")
print(f"   [Hidden Framework Logic] Passing tasks from {researcher.role} to {writer.role}...")
print(f"   [Result] 'Mars is red and dusty. Read all about it!'")
```

## 2. Low-Level Orchestration (PydanticAI & LangGraph)
SOTA architectures require explicitly typed contracts and state management. You define the exact inputs and outputs.

```python
from pydantic import BaseModel, Field

# Using Pydantic to enforce the contract between the Researcher and the Writer
class ResearchPayload(BaseModel):
    planet_name: str
    key_facts: list[str] = Field(description="Must contain exactly 3 facts.")

def explicit_research_node() -> ResearchPayload:
    print("🔬 Researcher Node: Executing strict schema lookup...")
    return ResearchPayload(planet_name="Mars", key_facts=["It is red.", "It has a thin atmosphere.", "It has two moons."])

def explicit_writer_node(data: ResearchPayload):
    print("✍️ Writer Node: Receiving typed data and drafting...")
    return f"Blog Draft: {data.planet_name} has {len(data.key_facts)} interesting facts..."

print("⚙️ Initiating Explicit Graph Framework...")
payload = explicit_research_node()
draft = explicit_writer_node(payload)
print(f"   [Result] {draft}")
```
