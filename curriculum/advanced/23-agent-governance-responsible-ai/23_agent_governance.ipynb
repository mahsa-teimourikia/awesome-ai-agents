## Part 1 — Agent Bill of Materials (AI BOM)

An AI BOM records the models, tools, and data access an agent has, allowing compliance audits.

```python
from dataclasses import dataclass
import json

@dataclass
class AIBOM:
    agent_name: str
    version: str
    foundation_models: list[str]
    allowed_tools: list[str]
    data_classification: str

bom = AIBOM(
    agent_name="HR-Assistant-Agent",
    version="1.2.0",
    foundation_models=["gpt-4o", "text-embedding-3-small"],
    allowed_tools=["read_employee_directory", "query_pto_balance"],
    data_classification="CONFIDENTIAL",
)

print("📋  AI BOM Generation Demo")
print("=" * 60)
print(json.dumps(bom.__dict__, indent=2))
print("\n  This artifact must be filed with the AI Governance Board before deployment.")
```
