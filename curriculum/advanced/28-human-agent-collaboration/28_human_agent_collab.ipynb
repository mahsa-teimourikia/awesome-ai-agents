## Part 1 — Handoff Packets

When pausing for human review, the agent must present a structured 'Handoff Packet' containing exactly what it wants to do and why.

```python
from dataclasses import dataclass

@dataclass
class HandoffPacket:
    proposed_action: str
    justification: str
    evidence_ids: list[str]
    confidence: str

def request_human_approval(packet: HandoffPacket) -> bool:
    print(f"  [Orchestrator] 🛑 Execution Paused. Human review required.")
    print(f"    Action : {packet.proposed_action}")
    print(f"    Why    : {packet.justification}")
    print(f"    Sources: {packet.evidence_ids}")
    print(f"    Conf.  : {packet.confidence}")
    return True  # Simulate human clicking 'Approve'

packet = HandoffPacket(
    proposed_action="DELETE FROM users WHERE last_login < '2020-01-01'",
    justification="Data retention policy requires purging accounts inactive >4 years.",
    evidence_ids=["policy-doc-42", "db-query-11"],
    confidence="HIGH"
)

print("🤝  Human Handoff Demo")
print("=" * 60)
approved = request_human_approval(packet)
print(f"\n  [Human] Status: {'APPROVED' if approved else 'REJECTED'}")
```
