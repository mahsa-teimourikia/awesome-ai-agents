## Part 1 — OpenTelemetry Traces

Agents must emit structured trace data (like OTel) for every LLM and tool invocation.

```python
import json, time

class OTelTracer:
    def __init__(self, trace_id: str):
        self.trace_id = trace_id
        self.spans = []
        
    def add_span(self, name: str, attributes: dict, duration_ms: int):
        span = {
            "trace_id": self.trace_id,
            "name": name,
            "duration_ms": duration_ms,
            "attributes": attributes
        }
        self.spans.append(span)
        
    def print_trace(self):
        print(json.dumps(self.spans, indent=2))

tracer = OTelTracer("trace-4bf92f")
tracer.add_span("chat.completions.create", {
    "llm.model": "gpt-4o",
    "llm.usage.prompt_tokens": 150,
    "llm.usage.completion_tokens": 45
}, 850)
tracer.add_span("tool.query_metrics", {
    "tool.name": "query_metrics",
    "tool.status": "OK"
}, 120)

print("🔭  Agent Observability Trace Demo")
print("=" * 60)
tracer.print_trace()
```
