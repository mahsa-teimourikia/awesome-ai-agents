## Part 1 — Protocol Boundary Selection

MCP is for tools. Agent Protocol is for UI-to-Agent. A2A is for Agent-to-Agent delegation.

```python
from dataclasses import dataclass

@dataclass
class ProtocolContract:
    name: str
    boundary: str
    purpose: str

protocols = [
    ProtocolContract("MCP", "Agent ↔ Tool/Resource", "Expose read/write functions and data to the LLM"),
    ProtocolContract("Agent Protocol", "Client/UI ↔ Agent", "Start, monitor, and retrieve artifacts from an agent run"),
    ProtocolContract("A2A", "Agent ↔ Agent", "Delegate tasks to remote specialized agents"),
]

print("🌐  Agent Protocol Stack")
print("=" * 60)
for p in protocols:
    print(f"  [{p.name}]")
    print(f"    Boundary : {p.boundary}")
    print(f"    Purpose  : {p.purpose}\n")
```
