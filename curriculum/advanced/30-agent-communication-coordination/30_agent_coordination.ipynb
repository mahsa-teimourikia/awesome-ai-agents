## Part 1 — The Blackboard Pattern

Agents write to a shared state (Blackboard) rather than chatting back and forth, reducing token overhead and maintaining a single source of truth.

```python
class Blackboard:
    def __init__(self):
        self.findings = []
        
    def write(self, agent: str, finding: str):
        self.findings.append({"agent": agent, "finding": finding})
        print(f"  [Blackboard] ✍️ {agent} wrote: '{finding}'")
        
    def read_all(self):
        return self.findings

bb = Blackboard()

print("📝  Blackboard Coordination Demo")
print("=" * 60)

bb.write("MetricsAgent", "Error rate spiked to 31% at 09:05Z")
bb.write("DeployAgent", "Deploy v2.1 finished at 09:04Z")

print("\n  [SupervisorAgent] Reading Blackboard to synthesize...")
for entry in bb.read_all():
    print(f"    - {entry['finding']} (via {entry['agent']})")
print("  [SupervisorAgent] Conclusion: Deploy v2.1 caused the spike.")
```
