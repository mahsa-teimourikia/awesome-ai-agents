## Part 1 — DAGs vs State Machines

DAGs are for strict dependency pipelines. State Machines (like LangGraph) allow for loops and autonomous retries.

```python
class Orchestrator:
    def __init__(self):
        self.state = "INIT"
        
    def step(self):
        if self.state == "INIT":
            print("  [Step] Gathering initial context...")
            self.state = "LLM_REASONING"
        elif self.state == "LLM_REASONING":
            print("  [Step] LLM reasoning... decided to use a tool.")
            self.state = "TOOL_EXECUTION"
        elif self.state == "TOOL_EXECUTION":
            print("  [Step] Tool executed. Result: data found. Returning to LLM...")
            self.state = "LLM_REASONING_2"
        elif self.state == "LLM_REASONING_2":
            print("  [Step] LLM reasoning... task complete.")
            self.state = "END"

orch = Orchestrator()
print("⚙️  State Machine Orchestration Demo")
print("=" * 60)
while orch.state != "END":
    orch.step()
print("  [Orchestrator] Workflow completed.")
```
