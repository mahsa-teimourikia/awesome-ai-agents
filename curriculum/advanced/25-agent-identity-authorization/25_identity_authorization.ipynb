## Part 1 — OAuth Delegation

Agents acting on behalf of a human user must inherit their permissions via delegated tokens, NOT use a super-user service account.

```python
from dataclasses import dataclass

@dataclass
class Token:
    subject: str
    scopes: list[str]

class ResourceServer:
    def access_file(self, token: Token, filename: str) -> str:
        if "drive.readonly" not in token.scopes:
            return "403 Forbidden: Missing drive.readonly scope"
        return f"200 OK: Contents of {filename}"

print("🔑  OAuth Delegation Demo")
print("=" * 60)

server = ResourceServer()

user_token = Token("alice@corp.com", ["drive.readonly", "email"])
agent_token = Token("agent-v1", ["drive.readonly"])  # Delegated subset

print(f"  User Access : {server.access_file(user_token, 'budget.pdf')}")
print(f"  Agent Access: {server.access_file(agent_token, 'budget.pdf')}")

rogue_agent_token = Token("agent-v2", ["email"])
print(f"  Rogue Agent : {server.access_file(rogue_agent_token, 'budget.pdf')}")
```
